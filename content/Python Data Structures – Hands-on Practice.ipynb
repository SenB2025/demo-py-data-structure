## List Manipulations

Lists can be modified after they are created. We can add new items, remove existing items, and change the order of items.

### Adding an item with `append()`

The `append()` method adds a new item to the **end** of a list.

```python
fruits = ["apple", "grapes", "orange", "mango"]

fruits.append("banana")

print(fruits)
```

## Removing an Item with `remove()`

The `remove()` method removes the **first matching item** from a list.

For example, we can remove `"orange"` from our `fruits` list.

```python
fruits = ["apple", "grapes", "orange", "mango", "banana"]

fruits.remove("orange")

print(fruits)
```

## Inserting an Item with `insert()`

The `insert()` method adds an item at a **specific position** in a list.

It takes two arguments:

`insert(index, item)`

For example, `insert(1, "banana")` places `"banana"` at index `1`.

```python
fruits = ["apple", "grapes", "mango", "banana"]

fruits.insert(1, "orange")

print(fruits)
```

## Removing an Item with `pop()`

The `pop()` method removes an item from a list and **returns the removed item**.

If no index is provided, `pop()` removes the last item.

```python
fruits = ["apple", "orange", "watermelon", "grapes", "mango"]

removed_fruit = fruits.pop()

print("Removed:", removed_fruit)
print("Remaining:", fruits)
```

## Sorting a List

The `sort()` method arranges the items in a list in ascending order.

For text, this normally means alphabetical order.

```python
fruits = ["mango", "apple", "orange", "banana"]

fruits.sort()

print(fruits)
```
